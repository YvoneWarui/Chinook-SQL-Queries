# **3. Generating a list of foreign keys(child table) where they are referenced from in the parent table(primary key)**

```python
%%sql
SELECT 
    CONSTRAINT_NAME AS 'Constraint Name',
    TABLE_NAME AS 'Child Table (Foreign Key)',
    COLUMN_NAME AS 'Child Column',
    REFERENCED_TABLE_NAME AS 'Parent Table (Primary Key)',
    REFERENCED_COLUMN_NAME AS 'Parent Column'
FROM
    information_schema.KEY_COLUMN_USAGE
WHERE
    TABLE_SCHEMA = 'chinook' 
    AND REFERENCED_TABLE_NAME IS NOT NULL;
```

# **4. Counting the number of rows and columns from the tables**
- used **JOIN** to join two schemas i.e the TABLES and COLUMN schema to get the row count and the column count

```python
%%sql
SELECT 
    t.TABLE_NAME AS 'Table Name',
    t.TABLE_ROWS AS 'Row Count (approx)',
    COUNT(c.COLUMN_NAME) AS 'Column Count'
FROM 
    information_schema.TABLES t
JOIN 
    information_schema.COLUMNS c 
    ON t.TABLE_SCHEMA = c.TABLE_SCHEMA 
    AND t.TABLE_NAME = c.TABLE_NAME
WHERE 
    t.TABLE_SCHEMA = 'chinook'
GROUP BY 
    t.TABLE_NAME, 
    t.TABLE_ROWS
ORDER BY 
    t.TABLE_ROWS DESC;
```

# 1: Filtering and Sorting (SELECT, WHERE, ORDER BY)

```python
%%sql
    -- This query finds all tracks that belong to Album ID 1 that are longer than 4 minutes, sorted from longest to shortest.
SELECT 
    TrackID, 
    Name, 
    AlbumID, 
    Milliseconds
FROM 
    Track
WHERE 
    AlbumID = 1
    AND 
    Milliseconds > 240000
ORDER BY 
    Milliseconds DESC;
```

```python
%%sql
    -- This query finds all tracks that have a UnitPrice of exactly $0.99 and are shorter than 3 minutes (180,000 milliseconds)   
SELECT 
    TrackID,
    Name,
    UnitPrice,
    Milliseconds
FROM 
    track
WHERE 
    UnitPrice = 0.99
    AND Milliseconds < 180000
ORDER BY
    milliseconds ASC
    LIMIT 10;
```

```python
%%sql
-- This query finds all customers who live in Brazil.
SELECT 
    CustomerID, 
    FirstName,
    LastName,
    Country
FROM 
    customer
WHERE Country = "Brazil";
```

# 2: Simple Aggregations (SUM, AVG, COUNT)

```python
%%sql
    /* This query finds out how much money the store has made in total sales, the average amount spent per invoice, 
    and the total number of individual orders placed.*/
SELECT 
    SUM(Total) AS 'Total sales',
    AVG(Total) AS 'Average Sales',
    COUNT(InvoiceID) AS 'Orders placed'
FROM 
    invoice;
```

```python
%%sql
   /* This query calculates the total revenue (SUM), the average invoice amount (AVG), and the 
    highest single purchase (MAX) out of all invoices billed to the USA (BillingCountry = 'USA').*/
SELECT
    BillingCountry,
    SUM(Total) AS 'Total_revenue',
    AVG(Total) AS 'Invoice amount',
    MAX(Total) AS 'Highest_Purchase'
FROM 
    invoice
WHERE BillingCountry = "USA";
```

# 3: Grouping Data (GROUP BY)

```python
%%sql
    -- This query calculates the total revenue generated by each country, ordered from our highest-earning market down to our lowest.
SELECT 
    BillingCountry,
    SUM(Total) AS 'Total_revenue',
    COUNT(InvoiceID) AS 'Orders made'
FROM 
    Invoice
GROUP BY 
    BillingCountry
ORDER BY
    Total_revenue DESC;
```

```python
%%sql
-- This query calculates the total revenue generated by each individual city in Canada
SELECT
    BillingCountry,
    BillingCity,
    SUM(Total) AS 'Total_revenue'
FROM 
    invoice
WHERE 
    BillingCountry = "Canada"
GROUP BY 
    billingcity
ORDER BY 
    Total_revenue DESC;
```

# 4: Filtering Groups (HAVING)

```python
%%sql
    -- This query finds only the top-tier customers who have spent a combined total of more than $45 across all of their orders.
SELECT 
    CustomerID,
    SUM(Total) AS 'Total_Revenue',
    COUNT(InvoiceID) AS 'Orders made'
FROM 
    Invoice
GROUP BY
    CustomerID
HAVING 
    Total_revenue > 45
ORDER BY 
    Total_Revenue DESC;
```

```python
%%sql
   /* This query finds the corporate accounts or bulk-buyers by grouping the records in the Invoice 
    table by CustomerId to find customers who have placed more than 6 individual orders total.*/
SELECT 
    CustomerID,
    Count(InvoiceID) AS 'orders'
FROM 
    invoice
GROUP BY
    CustomerID
HAVING 
    orders > 6
ORDER BY orders DESC;
```

# Complex Challenge 1: The High-Value Global Market Analysis
The Business Scenario: The marketing department wants to identify which non-US countries are our most reliable, high-value markets. They want to ignore small, one-off purchases and focus purely on countries where the average transaction size is solid, and the total revenue is substantial.

Your Task: Write a single query that satisfies these conditions:

Look at all invoices outside the USA (BillingCountry != 'USA').

Group the data by BillingCountry.

Calculate three metrics for each country: Total Revenue, Average Invoice Amount, and the Total Number of Orders.

Filter the final grouped results to only show countries where the Total Revenue is greater than $40.

Sort the output so that the country with the highest average invoice amount appears first.

```python
%%sql
SELECT
    BillingCountry,
    SUM(Total) AS 'Total_Revenue',
    AVG(Total) AS 'Average invoice amount',
    COUNT(InvoiceID) AS 'Total Orders'
FROM
    invoice
WHERE 
    BillingCountry != 'USA'  
GROUP BY 
    BillingCountry
HAVING 
    SUM(Total) > 40 
ORDER BY 
    AVG(Total) DESC;
```

# Complex Challenge 2: Employee Performance Check (Fraud & Outlier Detection)
The Business Scenario: Management wants to check the consistency of our order values. They are looking for individual customer profiles where a single, massive order might be distorting that customer's lifetime value statistics.

Your Task: Write a single query to flag these specific profiles:

Group all transactions in the Invoice table by CustomerId.

Calculate the Highest Single Purchase (MAX) and the Average Purchase (AVG) for each customer.

Filter the groups using a HAVING clause to only include customers whose highest single purchase is more than 3 times larger than their average purchase.

Sort the results from the highest maximum purchase down to the lowest.

```python
%%sql
SELECT
    CustomerID,
    AVG(Total) AS 'Average Purchase',
    MAX(Total) AS 'Highest Single Purchase'
FROM invoice
GROUP BY CustomerID
HAVING  MAX(Total) > 3*AVG(Total)
ORDER BY MAX(Total) DESC;
```

# **1.JOINS(Inner Join,Left Join)**

```python
%%sql
    -- This query links Customers with their dedicated Support Representative (Employee) to see who manages who
SELECT 
    c.CustomerId,
    CONCAT(c.FirstName, ' ', c.LastName) AS 'Customer Name',
    c.City AS 'Customer City',
    CONCAT(e.FirstName, ' ', e.LastName) AS 'Support Rep Name',
    e.Title AS 'Rep Title'
FROM 
    Customer c
INNER JOIN 
    Employee e ON e.EmployeeId = c.SupportRepId
ORDER BY 
    c.CustomerId
    LIMIT 10;
```

```python
%%sql 
    -- This query lists the top 10 longest tracks along with their Album Title and the Artist Name.
SELECT 
    t.Name AS 'Track Name',
    t.milliseconds,
    a.Title AS 'Album Title',
    r.Name AS 'Artist Name'
FROM
    track t
INNER JOIN 
   album a ON t.albumid = a.albumid
INNER JOIN
    artist r ON r.artistID = a.artistID
ORDER BY 
    t.Milliseconds DESC
LIMIT 10;
```

```python
%%sql
    -- This query finds "dormant" customers who have registered in the system  but have never placed a single order.
SELECT
    c.CustomerID,
    CONCAT(c.FirstName, ' ',c.LastName) AS 'Customer Name',
    c.email
FROM 
    customer c
LEFT JOIN 
    invoice i ON i.CustomerID = c.CustomerId
WHERE i.InvoiceID IS NULL;
```

```python
%%sql
   /* This query verifies that every music genre listed in the database actually
    has tracks assigned to it, or if there are empty genres sitting in the catalog.*/
SELECT 
    g.GenreId,
    g.name AS 'Music Genre',
    COUNT(t.trackid) AS 'no of tracks'
FROM 
    track t
RIGHT JOIN 
   genre g ON t.genreid = g.genreid
GROUP BY 
    g.GenreId,
    g.name 
ORDER BY 'no of tracks' DESC
LIMIT 10;
```

```python
%%sql
    /* This query shows a list of individual item purchases, showing the track name, 
    the media type format (e.g., MPEG audio file, AAC audio file), and the price paid on the invoice.*/
SELECT 
    t.Name As 'track name',
    m.Name AS 'Media type',
    i.UnitPrice
FROM 
    track t
INNER JOIN 
    mediatype m ON m.mediatypeid = t.mediatypeid
INNER JOIN 
    invoiceline i ON i.trackid = t.trackid
LIMIT 10;
```

# **2. Subqueries**

```python
%%sql
    -- This query finds all tracks that are longer than the overall average length of all tracks in the store.
SELECT 
    TrackId, 
    Name, 
    Milliseconds / 1000 AS 'Duration (Seconds)'
FROM 
    Track
WHERE 
    Milliseconds > (
        SELECT AVG(Milliseconds) FROM Track
    )
ORDER BY 
    Milliseconds DESC
    LIMIT 10;
```

```python
%%sql
    -- This query finds all customers who have bought at least one track from the 'Rock' genre.
SELECT 
    CustomerId,
    FirstName,
    LastName,
    Email
FROM 
    Customer
WHERE 
    CustomerId IN (
        SELECT DISTINCT 
            i.CustomerId
        FROM 
            Invoice i
        INNER JOIN 
            InvoiceLine il ON i.InvoiceId = il.InvoiceId
        INNER JOIN 
            Track t ON il.TrackId = t.TrackId
        INNER JOIN 
            Genre g ON t.GenreId = g.GenreId
        WHERE 
            g.Name = 'Rock'
    );
```

```python
%%sql
    -- This query finds the average total spend per customer.
SELECT 
    AVG(CustomerTotalSpend) AS 'Average Spend Per Customer'
FROM (
    -- Calculates total spend per individual customer
    SELECT 
        CustomerId,
        SUM(Total) AS CustomerTotalSpend
    FROM 
        Invoice
    GROUP BY 
        CustomerId
) AS CustomerSpends;
```

# **3. Window Functions**

```python
%%sql
    /*This query displays the track Name, AlbumId, Milliseconds, and assign a rank
    to each track based on its duration within its album (longest track = Rank 1).*/
SELECT 
    AlbumId,
    Name AS 'Track Name',
    Milliseconds,
    ROW_NUMBER() OVER (
        PARTITION BY AlbumId 
        ORDER BY Milliseconds DESC
    ) AS Track_Length_Rank
FROM 
    Track
ORDER BY 
    AlbumId, 
    Track_Length_Rank;
```

```python
%%sql
    -- This query ranks all customers by their overall total spending across the entire business using DENSE_RANK()
SELECT 
    CustomerId,
    SUM(Total) AS Total_Spent,
    DENSE_RANK() OVER (
        ORDER BY SUM(Total) DESC
    ) AS Overall_Rank
FROM 
    Invoice
GROUP BY 
    CustomerId
ORDER BY 
    Overall_Rank;
```

# Business Problem Solving

# **1: Top-Performing Products & Customers**

```python
%%sql
SELECT 
    c.CustomerId,
    CONCAT(c.FirstName, ' ', c.LastName) AS 'Customer Name',
    c.Country,
    COUNT(i.InvoiceId) AS 'Total Orders Placed',
    ROUND(SUM(i.Total), 2) AS 'Lifetime Spend ($)'
FROM 
    Customer c
INNER JOIN 
    Invoice i ON c.CustomerId = i.CustomerId
GROUP BY 
    c.CustomerId, 
    c.FirstName, 
    c.LastName, 
    c.Country
ORDER BY 
    `Lifetime Spend ($)` DESC
LIMIT 5;
```

# **2: Revenue Trends Over Time**

```python
%%sql
SELECT 
    YEAR(InvoiceDate) AS 'Sales Year',
    MONTH(InvoiceDate) AS 'Sales Month',
    COUNT(InvoiceId) AS 'Monthly Orders',
    ROUND(SUM(Total), 2) AS 'Monthly Revenue ($)'
FROM 
    Invoice
GROUP BY 
    YEAR(InvoiceDate), 
    MONTH(InvoiceDate)
ORDER BY 
    'Sales Year' ASC, 
    'Sales Month' ASC;
```

# **3: Top-Performing Tracks by Revenue**

```python
%%sql
SELECT 
    t.TrackId,
    t.Name AS 'Track Name',
    ROUND(SUM(il.UnitPrice * il.Quantity), 2) AS Total_Sales_Amount
FROM 
    Track t
INNER JOIN 
    InvoiceLine il ON t.TrackId = il.TrackId
GROUP BY 
    t.TrackId, 
    t.Name
ORDER BY 
    Total_Sales_Amount DESC
LIMIT 5;
```

# **4: Yearly Country Growth Trend**

```python
%%sql
SELECT 
    YEAR(InvoiceDate) AS Sales_Year,
    ROUND(SUM(Total), 2) AS Total_Revenue
FROM 
    Invoice
WHERE 
    BillingCountry = 'USA'
GROUP BY 
    YEAR(InvoiceDate)
ORDER BY 
    Sales_Year ASC;
```
